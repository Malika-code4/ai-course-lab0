```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = """
Artificial Intelligence
Machine Learning
Deep Learning
Neural Networks
Python
Data Science
"""

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white'
).generate(text)

plt.figure(figsize=(10,5))
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
